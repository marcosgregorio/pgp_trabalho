```python
import pandas as pd
df_pronto_para_hdbscan = pd.read_parquet("sample_data/dataset_limpo.parquet")
df_pronto_para_hdbscan.columns = [col.lower() for col in df_pronto_para_hdbscan.columns]

dataframe_completo = pd.read_parquet("sample_data/dataset_limpo_antes_normalizacao.parquet")


# ==========================================
# CÉLULA 1: TREINAMENTO DO MODELO HDBSCAN
# ==========================================
# Nota: Nas versões mais recentes (>= 1.3), o scikit-learn já inclui o HDBSCAN nativamente.
from sklearn.cluster import HDBSCAN

# Definindo o parâmetro principal
# min_cluster_size: quantos registros, no mínimo, formam um padrão válido? 
# Se for 10, qualquer grupo com menos de 10 ocorrências será considerado ruído (-1).
tamanho_minimo = 15

# Instanciando o modelo (note que não há o parâmetro 'eps')
modelo_hdbscan = HDBSCAN(min_cluster_size=tamanho_minimo)

# Aplicando o HDBSCAN
clusters_hdb = modelo_hdbscan.fit_predict(df_pronto_para_hdbscan)

# Salvando no dataframe original
dataframe_completo['cluster_hdbscan'] = clusters_hdb

print("Distribuição dos pontos por Cluster (HDBSCAN):")
print("Nota: O cluster '-1' representa os Ruídos (outliers).")
print(dataframe_completo['cluster_hdbscan'].value_counts())
```

```python
# ==========================================
# CÉLULA 2: VISUALIZAÇÃO DOS CLUSTERS EM 2D (HDBSCAN)
# ==========================================
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
dados_2d = pca.fit_transform(df_pronto_para_hdbscan)

df_plot = dataframe_completo.copy()
df_plot['Eixo_X'] = dados_2d[:, 0]
df_plot['Eixo_Y'] = dados_2d[:, 1]

# Transformando a saída do HDBSCAN em texto para o gráfico
df_plot['cluster_hdbscan'] = df_plot['cluster_hdbscan'].astype(str)

plt.figure(figsize=(12, 8))
sns.set_style("whitegrid")

# Separando ruído (-1) dos clusters reais para pintar de cinza
df_ruido = df_plot[df_plot['cluster_hdbscan'] == '-1']
df_clusters = df_plot[df_plot['cluster_hdbscan'] != '-1']

# Plota os ruídos primeiro (em cinza e mais transparentes)
sns.scatterplot(
    data=df_ruido, x='Eixo_X', y='Eixo_Y', 
    color='lightgrey', alpha=0.3, s=30, label='Ruído (-1)'
)

# Plota os clusters encontrados por cima
sns.scatterplot(
    data=df_clusters, x='Eixo_X', y='Eixo_Y',
    hue='cluster_hdbscan', palette='tab10', 
    alpha=0.8, s=60
)
plt.title('Clusters de Ocorrências (HDBSCAN + PCA)', fontsize=16, pad=15)
plt.xlabel('Componente Principal 1', fontsize=12)
plt.ylabel('Componente Principal 2', fontsize=12)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title='Grupos (HDBSCAN)')
plt.tight_layout()
plt.show()
```
